```python
import subprocess
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt
```

```python
def write_post_processing_config(start_index = 0, stop_index = 0, prefix='./', identifier='sweep', attitude_origin=np.eye(3, dtype=float), pattern_prefix="", pattern_fname="", start_frequency=150e6, stop_frequency=1200e6, number_of_frequencies=501, output_prefix="", clean_temporary_files=False ):

    p = Path(prefix)
    pattern_path = Path(pattern_prefix)
    output_path = Path(output_prefix)

    md = dict()
    
    md['start_index'] = start_index
    md['stop_index'] = stop_index
    md['prefix'] = prefix
    md['identifier'] = str(p.joinpath(identifier))
    md['attitude_origin'] = attitude_origin
    md['pattern_fname']  = str(pattern_path.joinpath(pattern_fname))
    md['start_frequency'] = start_frequency
    md['stop_frequency'] = stop_frequency
    md['number_of_frequencies'] = number_of_frequencies
    md['output_filename'] = str(output_path.joinpath(f'{identifier}_postprocessing_results'))
    md['clean_temporary_files'] = clean_temporary_files
    
    json.dump(md, open(p.joinpath(f'{identifier}_postprocessing.json'), 'w'))
    
    return md
```
