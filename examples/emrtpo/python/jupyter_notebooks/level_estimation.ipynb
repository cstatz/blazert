```python
import numpy as np
```

```python
P_diffuse = 0.5
P_reflect = 0.5
P_transmit = 1 - P_reflect 

# refractive indices
n1 = np.sqrt(1)
n2 = np.sqrt(2.2)
n3 = np.sqrt(7.5)

# reflection coefficients (amplitude)
R1_local = ((n2 - n1)/(n2 + n1))**2
R2_local = ((n3 - n2)/(n3 + n2))**2

T1_local = 1. - R1_local

freq = 660e6
k1 = 2*np.pi * freq/299792458.


d1 = 0.66  # distance source to first plane
d2 = 6     # distance first plane to second plane

r1 = d1
r2 = d1 + n2 * d2
```

```python
print(R1_local)
print(R2_local)
print(T1_local)
```

# Burried Plane

```python
# reference first plane at source
R1 = (P_diffuse) * 1./(2.*r1) 
print(R1)
```

```python
# reflection from second plane evaluated at source
R2 = ((1. - P_diffuse) * P_transmit) * T1_local * 1./(2.*r2) * P_reflect * R2_local * P_diffuse * T1_local
print(R2)
```

```python
20*np.log10(np.abs(R2/R1))
```

# Burried Sphere (small)

```python
radius = 2.
rcs = np.pi*radius**2
```

```python
# reference first plane at source
R1 = P_diffuse * 1./(2.*r1)
print(R1)
```

```python
# reflection from the small sphere evaluated at source
R2 = (1. - P_diffuse) * T1_local * 1./(r2**2) * R2_local/rcs * P_diffuse
print(R2)
```

```python
10*np.log10(np.abs(R2/R1))
```
